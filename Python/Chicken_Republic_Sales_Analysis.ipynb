```python
import pandas as pd

df = pd.read_excel("c:\\Users\\Tomiwa\\Desktop\\chicken republic\\chicken_republic_lagos_sales (1).xlsx")

print(df.head())
print(df.shape)
print(df.info())
print(df.describe())
```

```python
print(df.isnull().sum())
```

```python
print(df.duplicated().sum())
```

```python
print(df.columns.tolist())
```

```python
df["Calculated Sales"] = df["Quantity Sold"] * df["Unit Price (NGN)"]

df["Sales Difference"] = df["Calculated Sales"] - df["Total Sales (NGN)"]

print(df["Sales Difference"].abs().max())
```

```python
print(df["Location"].unique())
print(df["Product Category"].unique())
print(df["Product"].unique())
```

```python
df["Profit Margin %"] = (
    df["Profit (NGN)"] / df["Total Sales (NGN)"]
) 
```

```python
df["Profit Margin %"] = (
    df["Profit (NGN)"] / df["Total Sales (NGN)"]
).round(4)
```

```python
print(df["Profit Margin %"].head())
```

```python
df["Day"] = df["Date"].dt.day
```

```python
df["Day Name"] = df["Date"].dt.day_name()
```

```python
df["Month"] = df["Date"].dt.month_name()
```

```python
df["Month Number"] = df["Date"].dt.month
```

```python
df["Year"] = df["Date"].dt.year
```

```python
df["Week"] = df["Date"].dt.isocalendar().week
```

```python
total_sales = df["Total Sales (NGN)"].sum()

print(total_sales)
```

```python
total_profit = df["Profit (NGN)"].sum()

print(total_profit)
```

```python
outlet_analysis = df.groupby("Location").agg(
    Sales=("Total Sales (NGN)", "sum"),
    Profit=("Profit (NGN)", "sum"),
    Quantity=("Quantity Sold", "sum")
    )

outlet_analysis["Profit Margin %"] = (
    outlet_analysis["Profit"] / 
    outlet_analysis["Sales"]
) * 100

print(outlet_analysis.sort_values("Profit", ascending=False))
```

```python
category_analysis = df.groupby("Product Category").agg(
    Sales=("Total Sales (NGN)", "sum"),
    Profit=("Profit (NGN)", "sum"),
    Quantity=("Quantity Sold", "sum")
)

print(category_analysis.sort_values("Sales", ascending=False))
```

```python
best_sellers = df.groupby("Product").agg(
    Quantity=("Quantity Sold", "sum"),
    Sales=("Total Sales (NGN)", "sum"),
    Profit=("Profit (NGN)", "sum")
).sort_values("Quantity", ascending=False)

print(best_sellers)
```

```python
daily_sales = df.groupby("Date").agg(
    Sales=("Total Sales (NGN)", "sum"),
    Profit=("Profit (NGN)", "sum"),
    Quantity=("Quantity Sold", "sum")
).reset_index()
```
